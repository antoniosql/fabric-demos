## 2. Cargar el ground truth


```python
import pandas as pd

df = pd.read_csv("/lakehouse/default/Files/frasohome/FS_TestSet_Metrico_Mixto.csv")
df_eval = df.rename(columns={"Question": "question", "Expected response": "expected_answer"})[
    ["question", "expected_answer"]
]
display(df_eval)
```

## 3. Ejecutar la evaluación


```python
from fabric.dataagent.evaluation import evaluate_data_agent

AGENT_NAME = "FraSoHome BI Agent"

evaluation_id = evaluate_data_agent(
    df_eval,
    data_agent_name=AGENT_NAME,
    table_name="frasohome_eval_bi",
    data_agent_stage="production",
)
print("Evaluation run:", evaluation_id)
```

## 4. Resultados


```python
from fabric.dataagent.evaluation import get_evaluation_summary, get_evaluation_details

display(get_evaluation_summary(table_name="frasohome_eval_bi", verbose=True))

# Detalle solo de las que fallan o quedan 'unclear'
display(get_evaluation_details(
    evaluation_id=evaluation_id,
    table_name="frasohome_eval_bi",
    get_all_rows=False,
    verbose=True,
))
```
